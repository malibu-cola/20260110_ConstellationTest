## 深層強化学習の簡単な振り返り 📚

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit1/RL_process_game.jpg" alt="強化学習のプロセス" width="100%">

最初のユニットで学んだことの簡単な復習をしましょう。

- 強化学習（RL）とは、**行動を通じて学習する計算論的アプローチ**です。私たちは、**試行錯誤を繰り返しながら環境と相互作用**し、フィードバックとして報酬（正または負）を受け取ることで環境から学習するエージェントを構築します。

- 強化学習エージェントの目標は、**期待累積報酬（期待リターンとも呼ばれます）を最大化すること**です。これは、あらゆる目標は期待累積報酬の最大化として記述できるという「報酬仮説」に基づいています。

- 強化学習のプロセスは、**状態、行動、報酬、次の状態という一連の要素を出力するループ**になっています。

- 期待累積報酬（期待リターン）を計算する際には、**報酬を割り引きます（割引率を適用します）**。遠い将来の報酬よりも予測しやすいため、より早い段階（ゲームの序盤など）で得られる報酬の方が、実際に得られる可能性が高いと考えられるからです。

- 強化学習の問題を解くには、**最適な方策（ポリシー）を見つける**必要があります。方策とはAIの「脳」にあたるもので、ある状態においてどのような行動をとるべきかを決定します。最適な方策とは、期待リターンを最大化するような行動を選択する方策のことです。

最適な方策を見つける方法には、**2つ**のアプローチがあります。

- **方策を直接学習させる**方法：方策ベースの手法（ポリシーベース手法）。

- 各状態でエージェントが得られる期待リターンを示す**価値関数を学習させ、その関数を用いて方策を定義する**方法：価値ベースの手法。

- 最後に「深層強化学習（Deep RL）」について触れました。これは、**とるべき行動（方策ベース）や状態の価値（価値ベース）を推定するために深層ニューラルネットワークを導入する**ものであり、それゆえに「ディープ（深層）」という名前が付けられています。

## Gymnasiumとその仕組みを理解する 🤖

🏋 強化学習（RL）の環境を提供するライブラリが「Gymnasium」です。
**深層強化学習において、Gymnasiumは頻繁に使用することになります。**

Gymnasiumは、[Farama Foundation](https://farama.org/)によって保守・管理されている、**Gymライブラリの新しいバージョン**です。

Gymnasiumライブラリは、主に以下の2つの機能を提供します。

- **強化学習（RL）環境を作成**するためのインターフェース
- **環境のコレクション**（gym-control、Atari、Box2Dなど）

具体的な例を見る前に、まずは強化学習のループ（プロセス）をおさらいしておきましょう。

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit1/RL_process_game.jpg" alt="強化学習のプロセス" width="100%">

各ステップでの動作：
- エージェントは**環境**から**状態 (S0)**を受け取ります（例：ゲームの最初のフレーム）。
- その**状態 (S0)**に基づいて、エージェントは**行動 (A0)**をとります（例：右に移動する）。
- 環境は**新しい状態 (S1)**に遷移します（例：新しいフレーム）。
- 環境はエージェントに**報酬 (R1)**を与えます（例：ゲームオーバーになっていないため、*正の報酬 +1*）。


Gymnasiumを使用する場合：

1️⃣ `gymnasium.make()` を使って環境を作成します。

2️⃣ `observation = env.reset()` で環境を初期状態にリセットします。

各ステップでの動作：

3️⃣ モデルを使って行動を決定します（この例ではランダムな行動をとります）。

4️⃣ `env.step(action)` を使用して環境内でその行動を実行し、以下を取得します：
- `observation`: 新しい状態 (st+1)
- `reward`: 行動実行後に得られる報酬
- `terminated`: エピソードが終了したかどうか（エージェントが終了状態に到達したか）を示すフラグ
- `truncated`: 新バージョンで導入されたフラグ。制限時間切れや、エージェントが環境の境界外に出た場合などを示します。
- `info`: 追加情報を提供する辞書（環境によって内容は異なります）。

詳細な説明はこちら 👉 https://gymnasium.farama.org/api/env/#gymnasium.Env.step

エピソードが終了した場合：
- `observation = env.reset()` で環境を初期状態にリセットします。

**例を見てみましょう！** コードを必ず確認してください。

````python
import gymnasium as gym

# First, we create our environment called LunarLander-v2
env = gym.make("LunarLander-v2")

# Then we reset this environment
observation, info = env.reset()

for _ in range(20):
  # Take a random action
  action = env.action_space.sample()
  print("Action taken:", action)

  # Do this action in the environment and get
  # next_state, reward, terminated, truncated and info
  observation, reward, terminated, truncated, info = env.step(action)

  # If the game is terminated (in our case we land, crashed) or truncated (timeout)
  if terminated or truncated:
      # Reset the environment
      print("Environment is reset")
      observation, info = env.reset()

env.close()
````

報酬関数（各タイムステップで報酬を与える関数） 💰：

各ステップの後に報酬が与えられます。エピソードの合計報酬は、**そのエピソード内の全ステップで得られた報酬の総和**となります。

各ステップにおける報酬は、以下のように変化します：

- 着陸機が着陸パッドに近いほど増加し、遠いほど減少します。
- 着陸機の移動速度が遅いほど増加し、速いほど減少します。
- 着陸機の傾きが大きい（水平でない）ほど減少します。
- 地面に接地している脚1本につき10ポイント増加します。
- 側面エンジンを噴射しているフレームごとに0.03ポイント減少します。
- メインエンジンを噴射しているフレームごとに0.3ポイント減少します。

エピソード終了時には、**クラッシュした場合は-100ポイント、安全に着陸した場合は+100ポイントの追加報酬**が与えられます。

この問題を解決するために、SB3の**PPO**を使用します。[PPO（Proximal Policy Optimization：近接方策最適化）は、本コースで学習する深層強化学習アルゴリズムの中でも、SOTA（State-of-the-Art：最先端）とされる手法の一つです](https://stable-baselines3.readthedocs.io/en/master/modules/ppo.html#example%5D)。

PPOは、以下の手法を組み合わせたものです。
- *価値ベースの強化学習手法*：ある状態と行動において**とるべき最も価値の高い行動**を示す、行動価値関数を学習する手法。
- *方策ベースの強化学習手法*：**行動の確率分布を提示する**方策（ポリシー）を学習する手法。

Stable-Baselines3のセットアップは簡単です：

1️⃣ **環境を作成**します（今回の例では、上記ですでに作成済みです）

2️⃣ **使用するモデルを定義し、そのインスタンスを作成**します（例：`model = PPO("MlpPolicy")`）

3️⃣ `model.learn` を使用して**エージェントを学習**させ、学習のタイムステップ数を指定します

```
# 環境の作成
env = gym.make('LunarLander-v2')

# エージェントのインスタンス化
model = PPO('MlpPolicy', env, verbose=1)
# エージェントの学習
model.learn(total_timesteps=int(2e5))
```

Loading a saved model from the Hub is really easy.

You go to https://huggingface.co/models?library=stable-baselines3 to see the list of all the Stable-baselines3 saved models.

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/notebooks/unit1/copy-id.png" alt="Copy-id"/>

2. Then we just need to use load_from_hub with:
- The repo_id
- The filename: the saved model inside the repo and its extension (*.zip)

Because the model I download from the Hub was trained with Gym (the former version of Gymnasium) we need to install shimmy a API conversion tool that will help us to run the environment correctly.

Shimmy Documentation: https://github.com/Farama-Foundation/Shimmy

````python
!pip install shimmy
````

````python
from huggingface_sb3 import load_from_hub
repo_id = "Classroom-workshop/assignment2-omar" # The repo_id
filename = "ppo-LunarLander-v2.zip" # The model filename.zip

# When the model was trained on Python 3.8 the pickle protocol is 5
# But Python 3.6, 3.7 use protocol 4
# In order to get compatibility we need to:
# 1. Install pickle5 (we done it at the beginning of the colab)
# 2. Create a custom empty object we pass as parameter to PPO.load()
custom_objects = {
            "learning_rate": 0.0,
            "lr_schedule": lambda _: 0.0,
            "clip_range": lambda _: 0.0,
}

checkpoint = load_from_hub(repo_id, filename)
model = PPO.load(checkpoint, custom_objects=custom_objects, print_system_info=True)
````

Let's evaluate this agent:

````python
#@title
eval_env = Monitor(gym.make("LunarLander-v2"))
mean_reward, std_reward = evaluate_policy(model, eval_env, n_eval_episodes=10, deterministic=True)
print(f"mean_reward={mean_reward:.2f} +/- {std_reward}")
````
